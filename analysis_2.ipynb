```python
def remove_outliers(df, cols):
    df_clean = df.copy()
    for col in cols:
        if df_clean[col].dtype in ['float64', 'int64', 'float32']:
            Q1 = df_clean[col].quantile(0.25)
            Q3 = df_clean[col].quantile(0.75)
            IQR = Q3 - Q1
            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR
            df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
    return df_clean
```

```python
import plotly.graph_objects as go

df_sankey = data2.groupby(
    ["NAME_CASH_LOAN_PURPOSE", "NAME_CONTRACT_STATUS"]
).size().reset_index(name="count")

# Create node lists
sources = df_sankey["NAME_CASH_LOAN_PURPOSE"].tolist()
targets = df_sankey["NAME_CONTRACT_STATUS"].tolist()
values  = df_sankey["count"].tolist()

labels = list(set(sources + targets))
label_to_index = {label: idx for idx, label in enumerate(labels)}

source_ids = [label_to_index[s] for s in sources]
target_ids = [label_to_index[t] for t in targets]

fig = go.Figure(data=[go.Sankey(
    node=dict(label=labels),
    link=dict(
        source=source_ids,
        target=target_ids,
        value=values
    )
)])

fig.update_layout(title_text="Flow: Loan Purpose → Loan Status")
fig.show()
```
